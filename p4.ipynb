```python
# --- 1. JALANKAN INI DULU (Alat Pemeriksa Jawaban) ---
import hashlib

_K = {
    'T1_baris': '50a0a04f2d',
    'T1_kolom': 'b17ef6d19c',
    'T2_kosong': '3a481e7283',
    'T2_kolom': 'ef2d127de3',
    'T3_rating': '0e17daca5f',
    'T4_jasa': '43b87f618c',
    'T5_dup': 'c2356069e9',
    'T5_dupid': '8241649609',
    'T6_kat': '4ec9599fc2',
    'T6_katbersih': 'ef2d127de3',
    'T7_rp': '1e472b39b1',
    'T8_jumlah': '35135aaa6c',
    'T8_diskon': '19581e27de',
    'T9_anggota': 'aae0212936',
    'T9_strip': '7c252ab334',
    'T10_bisnis': '44c8031cb0',
    'T12_bulan': '4b227777d4',
    'T12_rusak': '48ce32e8ec'
}

_P = {
    'T1_baris': 'Pakai df.shape[0] atau len(df).',
    'T1_kolom': 'Pakai df.shape[1].',
    'T2_kosong': 'df.isna().sum() memberi per kolom; jumlahkan lagi dengan .sum().',
    'T2_kolom': 'Hitung berapa kolom yang nilai isna().sum()-nya lebih dari 0.',
    'T3_rating': 'Bandingkan kolom rating_layanan dengan -999, lalu jumlahkan.',
    'T4_jasa': 'Saring baris berkategori Jasa, lalu .isna().mean()*100 pada berat_kg.',
    'T5_dup': 'df.duplicated().sum() menghitung baris yang SELURUH kolomnya sama.',
    'T5_dupid': 'Pakai df[\'id_transaksi\'].duplicated(keep=False).sum().',
    'T6_kat': 'Hitung nunique() pada kolom kategori apa adanya.',
    'T6_katbersih': 'Rapikan: .str.strip().str.title().',
    'T7_rp': 'Ubah ke teks dengan .astype(str), lalu .str.startswith(\'Rp\').',
    'T8_jumlah': 'Hitung baris dengan jumlah kurang dari atau sama dengan nol.',
    'T8_diskon': 'Gabungkan dua kondisi dengan |.',
    'T9_anggota': 'nunique() pada id_anggota apa adanya.',
    'T9_strip': 'nunique() setelah .str.strip().',
    'T10_bisnis': 'Bandinkan dengan rumus total bayar.',
    'T12_bulan': 'Pakai .dt.month.nunique().',
    'T12_rusak': 'Bandingkan hasil konversi.'
}

def _h(x):
    return hashlib.sha256(str(x).strip().lower().encode()).hexdigest()[:10]

def cek(tugas, jawaban):
    """Periksa jawaban Anda tanpa menampilkan kuncinya."""
    if tugas not in _K:
        print('Nama tugas tidak dikenal:', tugas); return
    if _h(jawaban) == _K[tugas]:
        print(f'BENAR   {tugas} = {jawaban}')
    else:
        print(f'BELUM   {tugas} = {jawaban}')
        print('        Petunjuk:', _P.get(tugas, '-'))

print('Alat pemeriksa siap!')
```

```python
import pandas as pd
import numpy as np
import os

# 1. Pastikan dataset mandiri dimuat
FILE_MANDIRI = 'transaksi_koperasi_2025.csv'
df_mandiri = pd.read_csv(FILE_MANDIRI)

print(f"Berhasil memuat dataset: {df_mandiri.shape[0]} baris, {df_mandiri.shape[1]} kolom\n")

# 2. Menyusun profil kualitas data mandiri (Bagian 10)
temuan_mandiri = []

def catat_mandiri(dimensi, kolom, masalah, jumlah, keparahan, rencana):
    temuan_mandiri.append({
        'dimensi'   : dimensi,
        'kolom'     : kolom,
        'masalah'   : masalah,
        'jumlah'    : int(jumlah),
        'persen'    : round(jumlah / len(df_mandiri) * 100, 2),
        'keparahan' : keparahan,
        'rencana'   : rencana,
    })

# Merekam temuan-temuan penting dari dataset koperasi mandiri
catat_mandiri('Kelengkapan', 'saldo_simpanan', 'Kosong, diduga MNAR', df_mandiri['saldo_simpanan'].isna().sum(), 'Tinggi', 'Jangan diimputasi sembarangan; verifikasi ke pengurus')
catat_mandiri('Kelengkapan', 'rating_layanan', 'Placeholder -999', (df_mandiri['rating_layanan'] == -999).sum(), 'Tinggi', 'Ubah menjadi NaN sebelum dihitung')
catat_mandiri('Konsistensi', 'kategori', 'Satu kategori ditulis dengan berbagai variasi penulisan', 18, 'Tinggi', 'Seragamkan kategori dan tangani sinonim secara manual')
catat_mandiri('Keunikan', 'id_transaksi', 'ID transaksi berulang (duplikat nyaris)', int(df_mandiri['id_transaksi'].duplicated(keep=False).sum()), 'Tinggi', 'Telusuri sistem pencatatan kasir yang mencatat ganda')
catat_mandiri('Validitas', 'jumlah', 'Nilai jumlah barang nol atau negatif', int((df_mandiri['jumlah'] <= 0).sum()), 'Tinggi', 'Telusuri apakah retur atau kesalahan input')
catat_mandiri('Tipe Data', 'total_bayar', 'Tersimpan sebagai teks berformat "Rp"', int(df_mandiri['total_bayar'].astype(str).str.startswith('Rp').sum()), 'Tinggi', 'Hapus awalan teks dan konversi ke numerik')

profil_mandiri = pd.DataFrame(temuan_mandiri).sort_values(['keparahan', 'jumlah'], ascending=[True, False])

# 3. Simpan ke dalam folder reports/
os.makedirs('reports', exist_ok=True)
profil_mandiri.to_csv('reports/profil_kualitas_data.csv', index=False)

print("Profil kualitas data berhasil disusun dan disimpan ke 'reports/profil_kualitas_data.csv'")
display(profil_mandiri)
```
